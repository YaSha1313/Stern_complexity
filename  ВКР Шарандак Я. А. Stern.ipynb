```python
def c_it(n, k, l, m):
    return (2 * l * m * binomial(k//2, m) + 2 * m * (n - k) * binomial(k//2, m)^2 * 2^(-l))
```

```python
def c_inv(n, k):
    return ((1/2) * (n - k)^3 + k * (n - k)^2)
```

```python
def c_decode(n, k):
    return (n^2 * log(n, 2) + k + (n - k)^2 * log(n - k, 2))
```

```python
def p_success(n, k, d, l, m):
    return ((binomial(d, 2*m) * binomial(n - d, k - 2*m) * binomial(2*m, m) * binomial(n - k - d +2*m, l)) / (4^m * binomial(n, k) * binomial(n - k, l)))
```

```python
p_inv = 0.29
def classic_Stern_cost (n, k, d, l, m):
    return ((c_inv(n, k) + c_it(n, k, l, m)) / (p_inv * p_success(n, k, d, l, m)))

def Grover_Stern_cost (n, k, d, l, m):
    return (pi/4 * sqrt(1/(p_inv * p_success(n, k, d, l, m)))*(c_decode(n, k) + c_inv(n, k) + c_it(n, k, l, m)))
```
